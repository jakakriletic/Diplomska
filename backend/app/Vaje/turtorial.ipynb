```python
def counter(*array):
    count = 0
    for element in array:
        count += 1
    return count

print(counter(1, 2, 3, 4, 5))
```

```python
import pandas as pd
listime = []
listocena = []


def insert(name, grade):
    listime.append(name)
    listocena.append(grade)
    return listime, listocena

vnos = int(input("Vnesite število vnosov: "))

while vnos > 0:
    name = input("Vnesite ime: ")
    grade = int(input("Vnesite oceno: "))
    insert(name, grade)
    vnos -=1
panda = pd.DataFrame({"Ime": listime, "Ocena": listocena})
print(panda)
```
